```python
# Exception handling is used to manage runtime errors
# and prevent a program from stopping unexpectedly.

# 1. Handling Division by Zero

try:
    num1 = 100
    num2 = 0
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")


# 2. Handling Invalid Input

try:
    age = int("twenty")
    print("Age:", age)

except ValueError:
    print("Error: Please enter a valid number.")


# 3. Using Else and Finally

try:
    number = 25
    result = number / 5

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Execution completed.")


# 4. Handling Multiple Exceptions

try:
    numbers = [10, 20, 30]
    index = int(input("Enter an index: "))
    print("Value:", numbers[index])

except ValueError:
    print("Error: Enter a valid integer.")

except IndexError:
    print("Error: Index is outside the list.")


# 5. Raising a Custom Exception

def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient account balance.")
    return balance - amount

try:
    remaining_balance = withdraw(5000, 6000)
    print("Remaining Balance:", remaining_balance)

except ValueError as error:
    print("Transaction Failed:", error)
```

```python
# Object-Oriented Programming (OOP) is a programming approach
# that organizes code using classes and objects.
# It helps make programs reusable, organized and easier to maintain.

# 1. Creating a Class and Object

class Student:
    def display_details(self, name, branch):
        print("Student Name:", name)
        print("Branch:", branch)

student1 = Student()
student1.display_details("Sandeep", "AI and Data Science")


# 2. Using Constructors

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Employee:", self.name)
        print("Salary:", self.salary)

employee1 = Employee("Rahul", 45000)
employee1.display()


# 3. Encapsulation

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)
account.deposit(2000)

print("Account Balance:", account.get_balance())


# 4. Inheritance

class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    def drive(self):
        print("Car is moving")

car = Car()
car.start()
car.drive()


# 5. Polymorphism

class Calculator:
    def calculate(self, a, b):
        return a + b

class Multiplier:
    def calculate(self, a, b):
        return a * b

obj1 = Calculator()
obj2 = Multiplier()

print("Addition:", obj1.calculate(10, 5))
print("Multiplication:", obj2.calculate(10, 5))
```

```python
# Classes act as blueprints for creating objects.
# Inheritance allows one class to reuse another class's features.
# Polymorphism allows the same method to behave differently
# depending on the object that calls it.

# 1. Creating a Class with Constructor

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

employee = Employee("Rahul", 45000)
employee.display()


# 2. Single Inheritance

class Person:
    def show_name(self):
        print("Employee Name: Sandeep")

class Developer(Person):
    def show_role(self):
        print("Role: Python Developer")

developer = Developer()
developer.show_name()
developer.show_role()


# 3. Multilevel Inheritance

class Organization:
    def company(self):
        print("Company: Infosys")

class Department(Organization):
    def department(self):
        print("Department: Technology")

class Team(Department):
    def team_name(self):
        print("Team: Software Development")

team = Team()
team.company()
team.department()
team.team_name()


# 4. Method Overriding

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

dog = Dog()
dog.sound()


# 5. Using super()

class Manager:
    def __init__(self, name):
        self.name = name

class ProjectManager(Manager):
    def __init__(self, name, project):
        super().__init__(name)
        self.project = project

    def display(self):
        print("Manager:", self.name)
        print("Project:", self.project)

manager = ProjectManager("Arjun", "Automation System")
manager.display()


# 6. Polymorphism Through Different Classes

class CreditCard:
    def payment(self, amount):
        print("Credit Card Payment:", amount)

class UPI:
    def payment(self, amount):
        print("UPI Payment:", amount)

class NetBanking:
    def payment(self, amount):
        print("Net Banking Payment:", amount)

payment_methods = [CreditCard(), UPI(), NetBanking()]

for method in payment_methods:
    method.payment(5000)
```

```python
# Iterators are used to access elements one at a time.
# Generators produce values one at a time using the yield keyword.
# They help process large amounts of data efficiently.

# 1. Creating an Iterator

numbers = [10, 20, 30, 40, 50]

iterator = iter(numbers)

print("First Value:", next(iterator))
print("Second Value:", next(iterator))
print("Third Value:", next(iterator))


# 2. Iterating Through a Collection

fruits = ["Apple", "Mango", "Orange"]

for fruit in iter(fruits):
    print(fruit)


# 3. Creating a Custom Iterator

class NumberIterator:
    def __init__(self, limit):
        self.current = 1
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        raise StopIteration

numbers = NumberIterator(5)

for number in numbers:
    print(number)


# 4. Creating a Generator

def generate_numbers(limit):
    for number in range(1, limit + 1):
        yield number

for value in generate_numbers(5):
    print(value)


# 5. Fibonacci Series Using a Generator

def fibonacci(limit):
    a, b = 0, 1

    for _ in range(limit):
        yield a
        a, b = b, a + b

print("Fibonacci Series:", list(fibonacci(8)))


# 6. Generator Expression

squares = (number ** 2 for number in range(1, 6))

print("Squares:", list(squares))
```

```python
# ============================================================
# MASTER PYTHON PROGRAMS
# 23 Programming Exercises
# Converted from the supplied Java programs
# ============================================================

from math import gcd


# ------------------------------------------------------------
# Helper Functions
# ------------------------------------------------------------

def read_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter a valid integer.")


def read_array():
    size = read_int("Enter array size: ")

    if size < 0:
        print("Array size cannot be negative.")
        return []

    values = []

    for i in range(size):
        values.append(read_int(f"Enter element {i + 1}: "))

    return values


def read_matrix(rows, columns, label="matrix"):
    matrix = []

    print(f"Enter elements of the {label}:")

    for i in range(rows):
        row = []

        for j in range(columns):
            row.append(read_int(f"Element [{i}][{j}]: "))

        matrix.append(row)

    return matrix


# ------------------------------------------------------------
# 1. Multiplication Table
# ------------------------------------------------------------

def multiplication_table():
    number = read_int("Enter a number: ")

    for i in range(1, 11):
        print(f"{number} x {i} = {number * i}")


# ------------------------------------------------------------
# 2. Sum of Odd Numbers in a Range
# ------------------------------------------------------------

def sum_of_odd_numbers():
    limit = read_int("Enter the upper limit: ")

    total = sum(range(1, limit + 1, 2)) if limit >= 1 else 0

    print("Sum of odd numbers:", total)


# ------------------------------------------------------------
# 3. Factorial of a Number
# ------------------------------------------------------------

def find_factorial():
    number = read_int("Enter a number: ")

    if number < 0:
        print("Factorial is not defined for negative numbers.")
        return

    result = 1

    for i in range(2, number + 1):
        result *= i

    print("Factorial:", result)


# ------------------------------------------------------------
# 4. Sum of Digits
# ------------------------------------------------------------

def sum_of_digits():
    number = abs(read_int("Enter a number: "))

    total = 0

    while number > 0:
        total += number % 10
        number //= 10

    print("Sum of digits:", total)


# ------------------------------------------------------------
# 5. LCM of Two Numbers
# ------------------------------------------------------------

def find_lcm():
    a = read_int("Enter first number: ")
    b = read_int("Enter second number: ")

    if a == 0 or b == 0:
        result = 0
    else:
        result = abs(a * b) // gcd(a, b)

    print("LCM:", result)


# ------------------------------------------------------------
# 6. HCF / GCD of Two Numbers
# ------------------------------------------------------------

def find_hcf():
    a = read_int("Enter first number: ")
    b = read_int("Enter second number: ")

    print("HCF:", gcd(a, b))


# ------------------------------------------------------------
# 7. Prime Number Check
# ------------------------------------------------------------

def check_prime_number():
    number = read_int("Enter a number: ")

    if number <= 1:
        print("Not a prime number")
        return

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            print("Not a prime number")
            return

    print("Prime number")


# ------------------------------------------------------------
# 8. Reverse a Number
# ------------------------------------------------------------

def reverse_number():
    number = read_int("Enter a number: ")

    sign = -1 if number < 0 else 1

    reversed_number = int(str(abs(number))[::-1])

    print("Reversed number:", sign * reversed_number)


# ------------------------------------------------------------
# 9. Fibonacci Sequence
# ------------------------------------------------------------

def fibonacci_sequence():
    limit = read_int("Enter the maximum value: ")

    if limit < 0:
        print("Please enter a non-negative number.")
        return

    first, second = 0, 1
    sequence = []

    while first <= limit:
        sequence.append(first)
        first, second = second, first + second

    print("Fibonacci Sequence:", *sequence)


# ------------------------------------------------------------
# 10. Armstrong Number Check
# ------------------------------------------------------------

def check_armstrong_number():
    number = read_int("Enter a number: ")

    if number < 0:
        print("Not an Armstrong number")
        return

    digits = str(number)
    power = len(digits)

    total = sum(int(digit) ** power for digit in digits)

    if total == number:
        print("Armstrong number")
    else:
        print("Not an Armstrong number")


# ------------------------------------------------------------
# 11. Palindrome Number Check
# ------------------------------------------------------------

def check_palindrome():
    number = read_int("Enter a number: ")

    original = str(number)
    reversed_value = original[::-1]

    if original == reversed_value:
        print("Palindrome number")
    else:
        print("Not a Palindrome")


# ------------------------------------------------------------
# 12. Sum of Array Elements
# ------------------------------------------------------------

def sum_of_array_elements():
    values = read_array()

    print("Sum of array elements:", sum(values))


# ------------------------------------------------------------
# 13. Maximum and Minimum in Array
# ------------------------------------------------------------

def find_max_min_in_array():
    values = read_array()

    if not values:
        print("Array is empty.")
        return

    print("Maximum element:", max(values))
    print("Minimum element:", min(values))


# ------------------------------------------------------------
# 14. Reverse an Array
# ------------------------------------------------------------

def reverse_array():
    values = read_array()

    print("Reversed array:", values[::-1])


# ------------------------------------------------------------
# 15. Sort an Array
# ------------------------------------------------------------

def sort_array():
    values = read_array()

    print("Sorted array:", sorted(values))


# ------------------------------------------------------------
# 16. Count Occurrences in an Array
# ------------------------------------------------------------

def count_occurrences():
    values = read_array()

    target = read_int("Enter the number to count: ")

    print(
        f"The number {target} appears "
        f"{values.count(target)} times in the array."
    )


# ------------------------------------------------------------
# 17. Remove an Element from an Array
# ------------------------------------------------------------

def remove_element_from_array():
    values = read_array()

    target = read_int("Enter the number to remove: ")

    updated = [value for value in values if value != target]

    print("Updated array:", updated)


# ------------------------------------------------------------
# 18. Merge Two Sorted Arrays
# ------------------------------------------------------------

def merge_sorted_arrays():
    print("Enter the first sorted array:")
    first = read_array()

    print("Enter the second sorted array:")
    second = read_array()

    merged = []

    i = 0
    j = 0

    while i < len(first) and j < len(second):

        if first[i] <= second[j]:
            merged.append(first[i])
            i += 1

        else:
            merged.append(second[j])
            j += 1

    merged.extend(first[i:])
    merged.extend(second[j:])

    print("Merged sorted array:", merged)


# ------------------------------------------------------------
# 19. Search Element in 2D Array
# ------------------------------------------------------------

def search_in_2d_array():
    rows = read_int("Enter number of rows: ")
    columns = read_int("Enter number of columns: ")

    if rows <= 0 or columns <= 0:
        print("Rows and columns must be positive.")
        return

    matrix = read_matrix(rows, columns)

    target = read_int("Enter the number to search: ")

    found = False

    for i in range(rows):
        for j in range(columns):

            if matrix[i][j] == target:
                print(f"Element found at position [{i}][{j}]")
                found = True

    if not found:
        print("Element not found in the matrix.")


# ------------------------------------------------------------
# 20. Sum and Average of 2D Array
# ------------------------------------------------------------

def sum_and_average_2d_array():
    rows = read_int("Enter number of rows: ")
    columns = read_int("Enter number of columns: ")

    if rows <= 0 or columns <= 0:
        print("Rows and columns must be positive.")
        return

    matrix = read_matrix(rows, columns)

    total = sum(sum(row) for row in matrix)

    average = total / (rows * columns)

    print("Sum:", total)
    print("Average:", average)


# ------------------------------------------------------------
# 21. Multiplication of Two 2D Arrays
# ------------------------------------------------------------

def multiply_2d_arrays():
    rows1 = read_int("Enter rows in first matrix: ")
    columns1 = read_int("Enter columns in first matrix: ")

    rows2 = read_int("Enter rows in second matrix: ")
    columns2 = read_int("Enter columns in second matrix: ")

    if min(rows1, columns1, rows2, columns2) <= 0:
        print("All matrix dimensions must be positive.")
        return

    if columns1 != rows2:
        print("Matrix multiplication not possible.")
        return

    first = read_matrix(rows1, columns1, "first matrix")
    second = read_matrix(rows2, columns2, "second matrix")

    result = [
        [0 for _ in range(columns2)]
        for _ in range(rows1)
    ]

    for i in range(rows1):
        for j in range(columns2):
            for k in range(columns1):
                result[i][j] += first[i][k] * second[k][j]

    print("Resultant matrix:")

    for row in result:
        print(row)


# ------------------------------------------------------------
# 22. Transpose of a 2D Array
# ------------------------------------------------------------

def transpose_matrix():
    rows = read_int("Enter number of rows: ")
    columns = read_int("Enter number of columns: ")

    if rows <= 0 or columns <= 0:
        print("Rows and columns must be positive.")
        return

    matrix = read_matrix(rows, columns)

    transposed = [
        [matrix[i][j] for i in range(rows)]
        for j in range(columns)
    ]

    print("Transpose of the matrix:")

    for row in transposed:
        print(row)


# ------------------------------------------------------------
# 23. Sum of Diagonal Elements in 2D Array
# ------------------------------------------------------------

def sum_diagonals_2d_array():
    size = read_int("Enter matrix size (n x n): ")

    if size <= 0:
        print("Matrix size must be positive.")
        return

    matrix = read_matrix(size, size)

    total = sum(matrix[i][i] for i in range(size))

    total += sum(
        matrix[i][size - 1 - i]
        for i in range(size)
        if i != size - 1 - i
    )

    print("Sum of diagonal elements:", total)


# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------

MENU = {
    1: ("Multiplication Table", multiplication_table),
    2: ("Sum of Odd Numbers in a Range", sum_of_odd_numbers),
    3: ("Factorial", find_factorial),
    4: ("Sum of Digits", sum_of_digits),
    5: ("LCM", find_lcm),
    6: ("HCF / GCD", find_hcf),
    7: ("Prime Number Check", check_prime_number),
    8: ("Reverse a Number", reverse_number),
    9: ("Fibonacci Sequence", fibonacci_sequence),
    10: ("Armstrong Number Check", check_armstrong_number),
    11: ("Palindrome Check", check_palindrome),
    12: ("Sum of Array Elements", sum_of_array_elements),
    13: ("Maximum and Minimum in Array", find_max_min_in_array),
    14: ("Reverse an Array", reverse_array),
    15: ("Sort an Array", sort_array),
    16: ("Count Occurrences in an Array", count_occurrences),
    17: ("Remove an Element from an Array", remove_element_from_array),
    18: ("Merge Two Sorted Arrays", merge_sorted_arrays),
    19: ("Search Element in 2D Array", search_in_2d_array),
    20: ("Sum and Average of 2D Array", sum_and_average_2d_array),
    21: ("Multiply Two 2D Arrays", multiply_2d_arrays),
    22: ("Transpose of a 2D Array", transpose_matrix),
    23: ("Sum of Diagonal Elements", sum_diagonals_2d_array)
}


def main():
    print("=" * 55)
    print("       WELCOME TO PYTHON CODING PRACTICE")
    print("=" * 55)

    while True:

        print("\nChoose a program:\n")

        for number, (title, _) in MENU.items():
            print(f"{number}. {title}")

        print("0. Exit")

        choice = read_int("\nEnter your choice: ")

        if choice == 0:
            print("Thank you for practicing Python!")
            break

        if choice in MENU:
            title, function = MENU[choice]

            print(f"\n--- {title} ---\n")

            function()

        else:
            print("Invalid choice. Please select between 0 and 23.")


# Run the complete program
if __name__ == "__main__":
    main()
```
